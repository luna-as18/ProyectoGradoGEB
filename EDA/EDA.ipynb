```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar estilo de gráficos siguiendo principios de Gestalt
plt.style.use('ggplot')

# Empresas con valores nulos
empresas_con_nulos = {
    'e2': ['ORIG_STD_VOLUME'],
    'e3': ['ORIG_STD_VOLUME', 'ORIG_TEMPERATURE', 'ORIG_RAW_VOLUME'],
    'e7': ['ORIG_STD_VOLUME', 'ORIG_TEMPERATURE', 'ORIG_PRESSURE', 'ORIG_RAW_VOLUME'],
    'e9': ['ORIG_STD_VOLUME', 'ORIG_TEMPERATURE', 'ORIG_PRESSURE', 'ORIG_RAW_VOLUME'],
    'e10': ['ORIG_STD_VOLUME', 'ORIG_TEMPERATURE', 'ORIG_PRESSURE', 'ORIG_RAW_VOLUME']
}

# Filtrar DataFrames de empresas con nulos
dataframes_con_nulos = [df for nombre, df in zip(nombres, dataframes) if nombre in empresas_con_nulos]
nombres_con_nulos = [nombre for nombre in nombres if nombre in empresas_con_nulos]

# Iterar sobre cada DataFrame y crear el gráfico de proporción de nulos
for nombre, df in zip(nombres_con_nulos, dataframes_con_nulos):
    # Calcular la proporción de valores nulos por variable
    missing_values = df.isnull().mean().reset_index()
    missing_values.columns = ['Variable', 'Proporción de Nulos']
    
    # Filtrar solo las variables con valores nulos
    missing_values = missing_values[missing_values['Proporción de Nulos'] > 0]

    # Crear gráfico de barras horizontales
    plt.figure(figsize=(10, 6))
    sns.barplot(
        data=missing_values,
        y="Variable",
        x="Proporción de Nulos",
        palette="Reds_r"  # Contraste para resaltar las variables con más nulos
    )

    # Añadir etiquetas y título
    plt.xlabel("Proporción de Valores Nulos", fontsize=12, fontweight='bold')
    plt.ylabel("Variable", fontsize=12, fontweight='bold')
    plt.title(f"Proporción de Valores Nulos en {nombre}", fontsize=14, fontweight='bold')

    # Mejorar claridad con líneas de referencia y eliminar bordes innecesarios
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['right'].set_visible(False)

    # Mostrar el gráfico
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Lista de DataFrames y sus nombres
dataframes = [e1, e2, e3, e4, e5, e6, e7, e8, e9, e10]
nombres = ['e1', 'e2', 'e3', 'e4', 'e5', 'e6', 'e7', 'e8', 'e9', 'e10']

# Variables clave para analizar en histogramas
variables = ["ORIG_STD_VOLUME", "STD_VOLUME", "ORIG_TEMPERATURE", "TEMPERATURE", 
             "PRESSURE", "ORIG_PRESSURE", "ORIG_RAW_VOLUME", "RAW_VOLUME"]

# Iterar sobre cada DataFrame y cada variable
for nombre, df in zip(nombres, dataframes):
    for var in variables:
        if var in df.columns:  # Verificar que la variable exista en el DataFrame
            plt.figure(figsize=(10, 6))
            
            # Histograma de la variable
            sns.histplot(df[var], bins=30, kde=True, color='skyblue', alpha=0.7)
            
            # Líneas de estadísticos clave
            plt.axvline(df[var].mean(), color='red', linestyle='dashed', linewidth=2, label='Media')
            plt.axvline(df[var].median(), color='blue', linestyle='dashed', linewidth=2, label='Mediana')
            plt.axvline(df[var].mode().values[0], color='black', linestyle='dashed', linewidth=4, label='Moda')
            plt.axvline(df[var].quantile(0.25), color='yellow', linestyle='dashed', linewidth=2, label='Q1 (25%)')
            plt.axvline(df[var].quantile(0.75), color='yellow', linestyle='dashed', linewidth=2, label='Q3 (75%)')

            # Configuración del gráfico
            plt.title(f"Distribución de {var} en {nombre}", fontsize=14, fontweight='bold')
            plt.xlabel(var, fontsize=12)
            plt.ylabel("Frecuencia", fontsize=12)
            plt.legend()
            plt.grid(alpha=0.3)

            plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

def freedman_diaconis_bins(x: pd.Series) -> int:
    """Calcula el número óptimo de bins usando la regla de Freedman-Diaconis con un límite superior."""
    IQR = x.quantile(0.75) - x.quantile(0.25)
    N = x.dropna().size  # Número de observaciones sin nulos
    
    if IQR == 0 or N == 0:  # Evitar divisiones por cero
        return 10  # Número arbitrario si no hay variabilidad
    
    bin_width = 2 * IQR / N ** (1 / 3)
    bins = int(np.ceil((x.max() - x.min()) / bin_width))  # Calcular bins
    
    return max(min(bins, 100), 10)  # Asegurar entre 10 y 100 bins


# Variables clave para analizar en histogramas
variables = ["ORIG_STD_VOLUME", "STD_VOLUME", "ORIG_TEMPERATURE", "TEMPERATURE", 
             "PRESSURE", "ORIG_PRESSURE", "ORIG_RAW_VOLUME", "RAW_VOLUME"]
# Lista de DataFrames y sus nombres
dataframes = [e1, e2, e3, e4, e5, e6, e7, e8, e9, e10]
nombres = ['e1', 'e2', 'e3', 'e4', 'e5', 'e6', 'e7', 'e8', 'e9', 'e10']

# Iterar sobre cada DataFrame y cada variable
for nombre, df in zip(nombres, dataframes):
    for var in variables:
        if var in df.columns and not df[var].isnull().all():  # Verificar que la variable exista y no esté completamente vacía
            plt.figure(figsize=(10, 6))
            
            # Calcular número óptimo de bins
            num_bins = freedman_diaconis_bins(df[var])
            
            # Histograma de la variable
            sns.histplot(df[var], bins=num_bins, kde=True, color='skyblue', alpha=0.7)
            
            # Líneas de estadísticos clave
            plt.axvline(df[var].mean(), color='red', linestyle='dashed', linewidth=2, label='Media')
            plt.axvline(df[var].median(), color='blue', linestyle='dashed', linewidth=2, label='Mediana')
            plt.axvline(df[var].mode().values[0], color='black', linestyle='dashed', linewidth=4, label='Moda')
            plt.axvline(df[var].quantile(0.25), color='yellow', linestyle='dashed', linewidth=2, label='Q1 (25%)')
            plt.axvline(df[var].quantile(0.75), color='yellow', linestyle='dashed', linewidth=2, label='Q3 (75%)')

            # Configuración del gráfico
            plt.title(f"Distribución de {var} en {nombre}", fontsize=14, fontweight='bold')
            plt.xlabel(var, fontsize=12)
            plt.ylabel("Frecuencia", fontsize=12)
            plt.legend()
            plt.grid(alpha=0.3)

            plt.show()
```

```python
# Crear un diccionario para almacenar las modas por empresa
modas_por_empresa = {}

# Lista de DataFrames y sus nombres
dataframes = [e1, e2, e3, e4, e5, e6, e7, e8, e9, e10]
nombres = ['e1', 'e2', 'e3', 'e4', 'e5', 'e6', 'e7', 'e8', 'e9', 'e10']

# Iterar sobre cada DataFrame y calcular la moda de cada variable
for nombre, df in zip(nombres, dataframes):
    modas = df.mode().iloc[0]  # Obtener la primera moda de cada columna
    modas_por_empresa[nombre] = modas

# Mostrar las modas por empresa
for empresa, modas in modas_por_empresa.items():
    print(f"Modas de la empresa {empresa}:\n{modas}\n")
```
