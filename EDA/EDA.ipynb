```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

def freedman_diaconis_bins(x: pd.Series) -> int:
    """Calcula el número óptimo de bins usando la regla de Freedman-Diaconis con un límite superior."""
    IQR = x.quantile(0.75) - x.quantile(0.25)
    N = x.dropna().size  # Número de observaciones sin nulos
    
    if IQR == 0 or N == 0:  # Evitar divisiones por cero
        return 10  # Número arbitrario si no hay variabilidad
    
    bin_width = 2 * IQR / N ** (1 / 3)
    bins = int(np.ceil((x.max() - x.min()) / bin_width))  # Calcular bins
    
    return max(min(bins, 100), 10)  # Asegurar entre 10 y 100 bins


# Variables clave para analizar en histogramas
variables = ["ORIG_STD_VOLUME", "STD_VOLUME", "ORIG_TEMPERATURE", "TEMPERATURE", 
             "PRESSURE", "ORIG_PRESSURE", "ORIG_RAW_VOLUME", "RAW_VOLUME"]
# Lista de DataFrames y sus nombres
dataframes = [e1, e2, e3, e4, e5, e6, e7, e8, e9, e10]
nombres = ['e1', 'e2', 'e3', 'e4', 'e5', 'e6', 'e7', 'e8', 'e9', 'e10']

# Iterar sobre cada DataFrame y cada variable
for nombre, df in zip(nombres, dataframes):
    for var in variables:
        if var in df.columns and not df[var].isnull().all():  # Verificar que la variable exista y no esté completamente vacía
            plt.figure(figsize=(10, 6))
            
            # Calcular número óptimo de bins
            num_bins = freedman_diaconis_bins(df[var])
            
            # Histograma de la variable
            sns.histplot(df[var], bins=num_bins, kde=True, color='skyblue', alpha=0.7)
            
            # Líneas de estadísticos clave
            plt.axvline(df[var].mean(), color='red', linestyle='dashed', linewidth=2, label='Media')
            plt.axvline(df[var].median(), color='blue', linestyle='dashed', linewidth=2, label='Mediana')
            plt.axvline(df[var].mode().values[0], color='black', linestyle='dashed', linewidth=4, label='Moda')
            plt.axvline(df[var].quantile(0.25), color='yellow', linestyle='dashed', linewidth=2, label='Q1 (25%)')
            plt.axvline(df[var].quantile(0.75), color='yellow', linestyle='dashed', linewidth=2, label='Q3 (75%)')

            # Configuración del gráfico
            plt.title(f"Distribución de {var} en {nombre}", fontsize=14, fontweight='bold')
            plt.xlabel(var, fontsize=12)
            plt.ylabel("Frecuencia", fontsize=12)
            plt.legend()
            plt.grid(alpha=0.3)

            plt.show()
```

```python
def freedman_diaconis_bins(x: pd.Series) -> int:
    """Calculate the optimal number of bins using the Freedman-Diaconis rule."""
    IQR = x.quantile(0.75) - x.quantile(0.25)
    N = x.dropna().size  # Number of non-null observations
    
    if IQR == 0 or N == 0:  # Avoid division by zero
        return 10  # Arbitrary number if no variability
    
    bin_width = 2 * IQR / N ** (1 / 3)
    bins = int(np.ceil((x.max() - x.min()) / bin_width))  # Calculate bins
    
    return max(min(bins, 100), 10)  # Ensure between 10 and 100 bins

# Variables to analyze
variables = df_combined.columns.drop(['EFFECTIVE_DATE', 'Empresa_ID'])

# Plot distribution for each variable
for var in variables:
    if var in df_combined.columns and not df_combined[var].isnull().all():  # Check if the variable exists and is not completely empty
        plt.figure(figsize=(10, 6))
        
        # Calculate optimal number of bins
        num_bins = freedman_diaconis_bins(df_combined[var])
        
        # Plot histogram
        sns.histplot(df_combined[var], bins=num_bins, kde=True, color='skyblue', alpha=0.7)
        
        # Add lines for key statistics
        plt.axvline(df_combined[var].mean(), color='red', linestyle='dashed', linewidth=2, label='Mean')
        plt.axvline(df_combined[var].median(), color='blue', linestyle='dashed', linewidth=2, label='Median')
        plt.axvline(df_combined[var].mode().values[0], color='black', linestyle='dashed', linewidth=4, label='Mode')
        plt.axvline(df_combined[var].quantile(0.25), color='yellow', linestyle='dashed', linewidth=2, label='Q1 (25%)')
        plt.axvline(df_combined[var].quantile(0.75), color='yellow', linestyle='dashed', linewidth=2, label='Q3 (75%)')

        # Configure plot
        plt.title(f"Distribution of {var} in df_combined", fontsize=14, fontweight='bold')
        plt.xlabel(var, fontsize=12)
        plt.ylabel("Frequency", fontsize=12)
        plt.legend()
        plt.grid(alpha=0.3)

        plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Lista de columnas a transformar con escala logarítmica
log_scale_columns = ["STD_VOLUME", "ORIG_STD_VOLUME"]

# Obtener lista de empresas únicas
empresas_unicas = df_combined['Empresa_ID'].unique()

# Crear un diccionario de colores asociado a cada empresa
palette = sns.color_palette("tab10", len(empresas_unicas))
color_dict = {empresa: color for empresa, color in zip(empresas_unicas, palette)}

# Iterar sobre cada columna numérica y generar ECDF
for column in df_combined.columns:
    if column not in ['EFFECTIVE_DATE', 'Empresa_ID']:  # Excluir columnas no numéricas
        plt.figure(figsize=(10, 6))

        # Dibujar ECDF para cada empresa de manera individual, asegurando etiquetas
        for empresa in empresas_unicas:
            subset = df_combined[df_combined['Empresa_ID'] == empresa]
            sns.ecdfplot(
                data=subset,
                x=column,
                color=color_dict[empresa],
                label=f"Empresa {empresa}"
            )

        # Configuración de títulos y etiquetas
        plt.title(f"ECDF de {column} por Empresa", fontsize=14, fontweight='bold')
        plt.xlabel(column, fontsize=12)
        plt.ylabel("Proporción Acumulada", fontsize=12)
        plt.xticks(fontsize=10)
        plt.yticks(fontsize=10)

        # Agregar la leyenda con los nombres correctos
        plt.legend(title="Empresa", bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)

        # Aplicar escala logarítmica si la columna es de volumen
        if column in log_scale_columns:
            plt.xscale("log")
            plt.title(f"ECDF de {column} por Empresa (Escala Log)", fontsize=14, fontweight='bold')

        # Mejorar el diseño eliminando líneas innecesarias
        plt.grid(True, linestyle="--", alpha=0.5)
        plt.gca().spines["top"].set_visible(False)
        plt.gca().spines["right"].set_visible(False)

        # Ajustar el espacio para evitar que la leyenda se corte
        plt.subplots_adjust(right=0.75)

        plt.show()
```
