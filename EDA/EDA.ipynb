```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

def freedman_diaconis_bins(x: pd.Series) -> int:
    """Calcula el número óptimo de bins usando la regla de Freedman-Diaconis con un límite superior."""
    IQR = x.quantile(0.75) - x.quantile(0.25)
    N = x.dropna().size  # Número de observaciones sin nulos
    
    if IQR == 0 or N == 0:  # Evitar divisiones por cero
        return 10  # Número arbitrario si no hay variabilidad
    
    bin_width = 2 * IQR / N ** (1 / 3)
    bins = int(np.ceil((x.max() - x.min()) / bin_width))  # Calcular bins
    
    return max(min(bins, 100), 10)  # Asegurar entre 10 y 100 bins


# Variables clave para analizar en histogramas
variables = ["ORIG_STD_VOLUME", "STD_VOLUME", "ORIG_TEMPERATURE", "TEMPERATURE", 
             "PRESSURE", "ORIG_PRESSURE", "ORIG_RAW_VOLUME", "RAW_VOLUME"]
# Lista de DataFrames y sus nombres
dataframes = [e1, e2, e3, e4, e5, e6, e7, e8, e9, e10]
nombres = ['e1', 'e2', 'e3', 'e4', 'e5', 'e6', 'e7', 'e8', 'e9', 'e10']

# Iterar sobre cada DataFrame y cada variable
for nombre, df in zip(nombres, dataframes):
    for var in variables:
        if var in df.columns and not df[var].isnull().all():  # Verificar que la variable exista y no esté completamente vacía
            plt.figure(figsize=(10, 6))
            
            # Calcular número óptimo de bins
            num_bins = freedman_diaconis_bins(df[var])
            
            # Histograma de la variable
            sns.histplot(df[var], bins=num_bins, kde=True, color='skyblue', alpha=0.7)
            
            # Líneas de estadísticos clave
            plt.axvline(df[var].mean(), color='red', linestyle='dashed', linewidth=2, label='Media')
            plt.axvline(df[var].median(), color='blue', linestyle='dashed', linewidth=2, label='Mediana')
            plt.axvline(df[var].mode().values[0], color='black', linestyle='dashed', linewidth=4, label='Moda')
            plt.axvline(df[var].quantile(0.25), color='yellow', linestyle='dashed', linewidth=2, label='Q1 (25%)')
            plt.axvline(df[var].quantile(0.75), color='yellow', linestyle='dashed', linewidth=2, label='Q3 (75%)')

            # Configuración del gráfico
            plt.title(f"Distribución de {var} en {nombre}", fontsize=14, fontweight='bold')
            plt.xlabel(var, fontsize=12)
            plt.ylabel("Frecuencia", fontsize=12)
            plt.legend()
            plt.grid(alpha=0.3)

            plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Copia del dataframe original para no modificar los datos originales
df_clean = df_combined.copy()

# Lista de columnas en las que eliminaremos valores atípicos
numeric_columns = ["STD_VOLUME", "ORIG_STD_VOLUME", "ORIG_TEMPERATURE", "TEMPERATURE", 
                   "PRESSURE", "ORIG_PRESSURE", "ORIG_RAW_VOLUME", "RAW_VOLUME"]

# Eliminar valores atípicos por empresa usando el método IQR
for empresa in df_clean["Empresa_ID"].unique():
    df_empresa = df_clean[df_clean["Empresa_ID"] == empresa]
    for col in numeric_columns:
        if col in df_empresa.columns:
            Q1 = df_empresa[col].quantile(0.25)
            Q3 = df_empresa[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            df_clean = df_clean[~((df_clean["Empresa_ID"] == empresa) & 
                                  ((df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)))]

# Lista de columnas a transformar con escala logarítmica
log_scale_columns = ["STD_VOLUME", "ORIG_STD_VOLUME"]

# Obtener lista de empresas únicas después de limpiar los outliers
empresas_unicas = df_clean['Empresa_ID'].unique()

# Crear un diccionario de colores asociado a cada empresa
palette = sns.color_palette("tab10", len(empresas_unicas))
color_dict = {empresa: color for empresa, color in zip(empresas_unicas, palette)}

# Iterar sobre cada columna numérica y generar ECDF
for column in numeric_columns:
    plt.figure(figsize=(10, 6))

    # Dibujar ECDF para cada empresa
    for empresa in empresas_unicas:
        subset = df_clean[df_clean['Empresa_ID'] == empresa]
        sns.ecdfplot(
            data=subset,
            x=column,
            color=color_dict[empresa],
            label=f"Empresa {empresa}"
        )

    # Configuración del gráfico
    plt.title(f"ECDF de {column} por Empresa (Sin Outliers)", fontsize=14, fontweight='bold')
    plt.xlabel(column, fontsize=12)
    plt.ylabel("Proporción Acumulada", fontsize=12)
    plt.legend(title="Empresa", bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)
    
    # Aplicar escala logarítmica si la columna es de volumen
    if column in log_scale_columns:
        plt.xscale("log")
        plt.title(f"ECDF de {column} por Empresa (Escala Log, Sin Outliers)", fontsize=14, fontweight='bold')

    # Ajustes visuales
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.gca().spines["top"].set_visible(False)
    plt.gca().spines["right"].set_visible(False)
    plt.subplots_adjust(right=0.75)

    plt.show()








```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Definir función para calcular el número óptimo de bins usando la regla de Freedman-Diaconis
def freedman_diaconis_bins(x: pd.Series) -> int:
    """Calcula el número óptimo de bins usando la regla de Freedman-Diaconis con un límite superior."""
    IQR = x.quantile(0.75) - x.quantile(0.25)
    N = x.dropna().size  # Número de observaciones sin nulos
    
    if IQR == 0 or N == 0:  # Evitar divisiones por cero
        return 10  # Número arbitrario si no hay variabilidad
    
    bin_width = 2 * IQR / N ** (1 / 3)
    bins = int(np.ceil((x.max() - x.min()) / bin_width))  # Calcular bins
    
    return max(min(bins, 100), 10)  # Asegurar entre 10 y 100 bins


# Variables a analizar
variables = ["ORIG_STD_VOLUME", "STD_VOLUME", "ORIG_TEMPERATURE", "TEMPERATURE", 
             "PRESSURE", "ORIG_PRESSURE", "ORIG_RAW_VOLUME", "RAW_VOLUME"]

# Lista de DataFrames y sus nombres
dataframes = [e1, e2, e3, e4, e5, e6, e7, e8, e9, e10]
nombres = ['e1', 'e2', 'e3', 'e4', 'e5', 'e6', 'e7', 'e8', 'e9', 'e10']

# Eliminar outliers de cada DataFrame
filtered_dataframes = []
for df in dataframes:
    df_filtered = df.copy()
    for var in variables:
        if var in df_filtered.columns and not df_filtered[var].isnull().all():
            Q1 = df_filtered[var].quantile(0.25)
            Q3 = df_filtered[var].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            df_filtered = df_filtered[(df_filtered[var] >= lower_bound) & (df_filtered[var] <= upper_bound)]
    filtered_dataframes.append(df_filtered)

# Crear gráficos organizados por variable
for var in variables:
    fig, axes = plt.subplots(2, 5, figsize=(18, 10), sharex=True, sharey=True)  # 2 filas, 5 columnas
    fig.suptitle(f"Distribución de {var} por empresa (Sin Outliers)", fontsize=16, fontweight='bold')

    for ax, df_filtered, nombre in zip(axes.flat, filtered_dataframes, nombres):
        if var in df_filtered.columns and not df_filtered[var].isnull().all():
            num_bins = freedman_diaconis_bins(df_filtered[var])  # Calcular bins óptimos
            sns.histplot(df_filtered[var], bins=num_bins, kde=True, color='skyblue', alpha=0.7, ax=ax)

            # Líneas de estadísticas clave
            ax.axvline(df_filtered[var].mean(), color='red', linestyle='dashed', linewidth=2, label='Media')
            ax.axvline(df_filtered[var].median(), color='blue', linestyle='dashed', linewidth=2, label='Mediana')
            ax.axvline(df_filtered[var].mode().values[0], color='black', linestyle='dashed', linewidth=4, label='Moda')
            ax.axvline(df_filtered[var].quantile(0.25), color='yellow', linestyle='dashed', linewidth=2, label='Q1 (25%)')
            ax.axvline(df_filtered[var].quantile(0.75), color='yellow', linestyle='dashed', linewidth=2, label='Q3 (75%)')

            ax.set_title(f"{nombre}", fontsize=12)
            ax.set_xlabel(var)
            ax.set_ylabel("Frecuencia")
            ax.grid(alpha=0.3)

    # Ajustar layout
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
..............
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Lista de DataFrames y sus nombres
dataframes = [e1, e2, e3, e4, e5, e6, e7, e8, e9, e10]
nombres = ['e1', 'e2', 'e3', 'e4', 'e5', 'e6', 'e7', 'e8', 'e9', 'e10']

# Variables a analizar
variables = ["ORIG_STD_VOLUME", "STD_VOLUME", "ORIG_TEMPERATURE", "TEMPERATURE", 
             "PRESSURE", "ORIG_PRESSURE", "ORIG_RAW_VOLUME", "RAW_VOLUME"]

# Crear un DataFrame combinado con identificador de empresa
df_combined = pd.concat([df.assign(Empresa_ID=name) for df, name in zip(dataframes, nombres)], ignore_index=True)

# Eliminar valores atípicos (outliers) por empresa y variable usando IQR
df_filtered = df_combined.copy()
for var in variables:
    for empresa in nombres:
        subset = df_combined[df_combined["Empresa_ID"] == empresa]
        
        # Calcular IQR
        Q1 = subset[var].quantile(0.25)
        Q3 = subset[var].quantile(0.75)
        IQR = Q3 - Q1
        
        # Definir umbrales para detectar outliers
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # Eliminar valores fuera del rango
        df_filtered = df_filtered[
            ~((df_filtered["Empresa_ID"] == empresa) & ((df_filtered[var] < lower_bound) | (df_filtered[var] > upper_bound)))
        ]

# Graficar boxplots SIN outliers
for var in variables:
    if var in df_filtered.columns:
        plt.figure(figsize=(12, 6))
        sns.boxplot(data=df_filtered, x="Empresa_ID", y=var, palette="viridis")
        plt.title(f"Boxplot de {var} por Empresa (Sin Outliers)")
        plt.xticks(rotation=45)
        plt.grid(alpha=0.3)
        plt.show()

# Graficar análisis temporal de la presión (si hay fechas)
if "EFFECTIVE_DATE" in df_filtered.columns:
    df_filtered["EFFECTIVE_DATE"] = pd.to_datetime(df_filtered["EFFECTIVE_DATE"], format='%Y-%m-%d %H:%M:%S', errors='coerce')

    plt.figure(figsize=(12, 6))
    sns.lineplot(data=df_filtered, x="EFFECTIVE_DATE", y="PRESSURE", hue="Empresa_ID", alpha=0.7, palette="tab10")
    plt.title("Evolución de la Presión por Empresa (Sin Outliers)")
    plt.xlabel("Fecha")
    plt.ylabel("Presión")
    plt.legend(title="Empresa", bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Lista de DataFrames y sus nombres
dataframes = [e1, e2, e3, e4, e5, e6, e7, e8, e9, e10]
nombres = ['e1', 'e2', 'e3', 'e4', 'e5', 'e6', 'e7', 'e8', 'e9', 'e10']

# 🔹 Definir función para graficar dispersión en subgráficos
def scatter_subplots(x_var, y_var, titulo):
    fig, axes = plt.subplots(2, 5, figsize=(18, 10))  # Crear una grilla de subgráficos 2x5
    axes = axes.flatten()  # Convertir en una lista plana para iterar más fácilmente

    for i, (df, name) in enumerate(zip(dataframes, nombres)):
        if x_var in df.columns and y_var in df.columns:  # Verificar existencia de variables
            sns.scatterplot(data=df, x=x_var, y=y_var, alpha=0.7, ax=axes[i])
            axes[i].set_title(f"{name}", fontsize=12, fontweight='bold')
            axes[i].set_xlabel(x_var)
            axes[i].set_ylabel(y_var)
            axes[i].grid(True, linestyle="--", alpha=0.5)

    # Configuración general
    fig.suptitle(titulo, fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()

# 📌 Graficar "ORIG_PRESSURE" vs "ORIG_STD_VOLUME" por empresa
scatter_subplots("ORIG_PRESSURE", "ORIG_STD_VOLUME", "Dispersión ORIG_PRESSURE vs ORIG_STD_VOLUME por Empresa")

# 📌 Graficar "ORIG_TEMPERATURE" vs "ORIG_STD_VOLUME" por empresa
scatter_subplots("ORIG_TEMPERATURE", "ORIG_STD_VOLUME", "Dispersión ORIG_TEMPERATURE vs ORIG_STD_VOLUME por Empresa")
```
